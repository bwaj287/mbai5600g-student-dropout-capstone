# Sentiment Analysis in Python: From Lexicon-Based to Transformer-Based Methods

## 1. What is Sentiment Analysis?

### Definition
Sentiment Analysis (or Opinion Mining) is a Natural Language Processing (NLP) task that identifies and classifies the emotional tone expressed in text, typically as:

* Positive
* Negative
* Neutral

### Why is it Important?
Sentiment analysis helps organizations understand opinions, attitudes, and emotions expressed in:
* Customer reviews
* Social media posts
* Surveys and feedback
* News articles
* Financial analysis 

### Example

**Text:** "The new smartphone is amazing and the battery lasts all day!"

**Sentiment:** POSITIVE

### Common Applications
* Brand monitoring
* Customer satisfaction analysis
* Social media analytics
* Product review mining
* Political opinion analysis
* Financial sentiment analysis

# 2. Lexicon-Based Sentiment Analysis

Lexicon-based methods rely on predefined dictionaries of words associated with sentiment scores.

Two popular Python libraries are:

1. TextBlob 
2. VADER (Valence Aware Dictionary and sEntiment Reasoner)

## 2.2 TextBlob

TextBlob is a high-level Python library built on top of NLTK and pattern that provides a simple API for common Natural Language Processing (NLP) tasks, including sentiment analysis. It is designed to be easy to use, making it suitable for **beginners** and **quick prototyping**.

For sentiment analysis, TextBlob uses a lexicon- and rule-based approach trained on **movie reviews**. It evaluates text **polarity** and **subjectivity** without requiring any custom training data.

TextBlob returns two key sentiment scores:

* **Polarity**: ranges from -1 (very negative) to +1 (very positive)
* **Subjectivity**: ranges from 0 (very objective/factual) to 1 (very subjective/opinion-based)

Unlike more rule-heavy systems such as VADER, TextBlob does not explicitly account for features like **capitalization**, **emojis**, or **punctuation** emphasis. Instead, it relies primarily on the **averaged polarity** of words and phrases in its internal lexicon.

Because of its simplicity and ease of use, TextBlob is commonly used for **baseline sentiment analysis**, educational purposes, and quick text polarity checks, though it may be less robust on informal or social media-style text compared to specialized tools.

The example below demonstrates how to use TextBlob to compute sentiment polarity and classify text as POSITIVE, NEGATIVE, or NEUTRAL based on the polarity score.

````python
from textblob import TextBlob

def get_sentiment(score):
    if score > 0.05:
        return "POSITIVE"
    elif score < -0.05:
        return "NEGATIVE"
    else:
        return "NEUTRAL"

examples = [
    "I love this product.",                          # base sentiment
    "I LOVE this product.",                          # capitalization
    "I love this product!!!",                        # punctuation emphasis
    "I really love this product.",                   # booster word
    "I absolutely love this product.",               # strong booster
    "I do not love this product.",                   # negation
    "The product is not good.",                        # negation + positive word
    "The product is great, but the battery is bad.",  # contrastive conjunction
    "The product is excellent 😊",                  # emoji
    "I REALLY LOVE this product!!! 😊",               # combined heuristics
    "This product is so great that it makes me puke!"         # sarcastic sentence
]

for text in examples:
    blob = TextBlob(text)
    score = blob.sentiment
    print(text)
    print(score, ", 'sentiment':", get_sentiment(score.polarity))
    print("-" * 100)
````

## 2.2 VADER

VADER (Valence Aware Dictionary and sEntiment Reasoner) is a **lexicon- and rule-based** sentiment analysis tool designed for **social media posts** and other forms of informal text. It uses a predefined sentiment lexicon to assign base sentiment scores to words and then applies a set of **linguistic rules to capture contextual cues commonly found in online communication**.

VADER generates four sentiment scores for a given text:

* **Positive (pos)**: proportion of positive sentiment
* **Negative (neg)**: proportion of negative sentiment
* **Neutral (neu)**: proportion of neutral sentiment
* **Compound**: normalized overall sentiment score ranging from **-1** (most negative) to **+1** (most positive)

Unlike simple lexicon-based approaches, VADER adjusts the base sentiment score using several rule-based heuristics, including:

* Capitalization (e.g., **LOVE** vs. *love*)
* Degree modifiers (e.g., *very*, *extremely*)
* Punctuation emphasis (e.g., *!!!*)
* Negations (e.g., *not good*)
* Contrastive conjunctions (e.g., *good, but expensive*)
* Emojis and emoticons

Because it requires no training data and is computationally efficient, VADER is widely used for sentiment analysis of tweets, online reviews, comments, and other user-generated content.

The example below demonstrates how to use the VADER sentiment analysis tool to evaluate the sentiment of multiple sentences, applying rule-based adjustments for factors such as capitalization, punctuation, negation, intensity modifiers, contrastive conjunctions, and emojis, and then classifying each sentence as POSITIVE, NEGATIVE, or NEUTRAL based on the compound sentiment score.

````python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

def get_sentiment(score):
    compound = score['compound']
    if compound >= 0.05:
        return "POSITIVE"
    elif compound <= -0.05:
        return "NEGATIVE"
    else:
        return "NEUTRAL"
        
analyzer = SentimentIntensityAnalyzer()

examples = [
    "I love this product.",                          # base sentiment
    "I LOVE this product.",                          # capitalization
    "I love this product!!!",                        # punctuation emphasis
    "I really love this product.",                   # booster word
    "I absolutely love this product.",               # strong booster
    "I do not love this product.",                   # negation
    "The product is not good.",                        # negation + positive word
    "The product is great, but the battery is bad.",  # contrastive conjunction
    "The product is excellent 😊",                  # emoji
    "I REALLY LOVE this product!!! 😊",               # combined heuristics
    "This product is so great that it makes me puke!"         # sarcastic sentence
]

for text in examples:
    score = analyzer.polarity_scores(text)
    print(text)
    print(score, ", 'sentiment':", get_sentiment(score))
    print("-" * 100)
````


# 2. Transformer-Based Sentiment Analysis - BERT

BERT (Bidirectional Encoder Representations from Transformers) is a deep learning-based language model developed for understanding the context of words in a sentence by considering both **left** and **right** context simultaneously. Unlike traditional lexicon-based methods, BERT is **pre-trained** on large text corpora using transformer architecture and can be fine-tuned for specific NLP tasks such as sentiment analysis, text classification, and question answering.

For sentiment analysis, BERT produces a contextual representation of an entire sentence and predicts sentiment labels (e.g., positive, negative, or neutral) based on learned patterns rather than predefined rules or dictionaries.

Unlike rule-based or lexicon-based approaches such as VADER, BERT captures complex linguistic phenomena such as:

* Word context dependence (same word can have different meanings)
* Sarcasm and implicit sentiment cues (to some extent)
* Long-range dependencies in text
* Subtle semantic relationships between words

Since BERT is a data-driven model trained on large-scale corpora, it typically achieves higher accuracy than traditional sentiment analysis methods, especially on complex or domain-specific text. However, it requires more computational resources and labeled training data for fine-tuning.

The example below demonstrates how to use a pretrained BERT-based model (Twitter RoBERTa Sentiment) for sentiment classification. The input sentences are first tokenized and then passed through the model, which outputs probabilities for three sentiment classes: POSITIVE, NEGATIVE, and NEUTRAL. The sentiment label is assigned based on the **class with the highest probability**.

Unlike VADER and TextBlob, which produce a continuous sentiment score (e.g., polarity ranging from -1 to +1), BERT-based models do not inherently compute polarity scores. Instead, they generate probability distributions across sentiment classes, where each probability represents the model’s **confidence that the input belongs to a particular sentiment category**.

````python
from transformers import pipeline

def format_sentiment(result):
    label_map = {
        "LABEL_0": "NEGATIVE",
        "LABEL_1": "NEUTRAL",
        "LABEL_2": "POSITIVE"
    }
    return f"{{'label': '{label_map[result['label']]}', 'Confidence': {result['score'] * 100:.2f}%}}"
    
# Load pretrained BERT sentiment analysis pipeline
classifier = pipeline(
    "sentiment-analysis",
    model = "cardiffnlp/twitter-roberta-base-sentiment"
)

examples = [
    "I love this product.",
    "I LOVE this product.",
    "I love this product!!!",
    "I really love this product.",
    "I absolutely love this product.",
    "I do not love this product.",
    "The product is not good.",
    "The product is great, but the battery is bad.",
    "The product is excellent 😊",
    "I REALLY LOVE this product!!! 😊",
    "This product is so great that it makes me puke!",
    "The product is available in different sizes and colors."
]

for text in examples:
    result = classifier(text)[0]
    print(text)
    print(format_sentiment(result))
    print("-" * 90)
````

# Summary

1. Sentiment Analysis is the process of identifying and classifying opinions, emotions, or attitudes in text as **positive**, **negative**, or **neutral**.

2. Lexicon-based methods such as **VADER** and **TextBlob** are computationally efficient, easy to implement, and require no labeled training data.

3. Lexicon-based approaches often struggle with contextual understanding, complex linguistic structures, domain-specific language, and sarcasm.

4. **BERT** leverages the Transformer architecture to capture contextual relationships between words, enabling a deeper understanding of language and generally achieving higher accuracy than traditional lexicon-based methods.

5. BERT performs well on complex and nuanced language but can still encounter difficulties with sarcasm, ambiguity, figurative language, and highly specialized domain terminology unless fine-tuned on relevant data.

6. For modern applications such as social media monitoring, customer feedback analysis, and review analytics, Transformer-based models such as **BERT**, **RoBERTa**, and other large language models are generally preferred due to their superior contextual understanding and predictive performance.

7. When a task requires a higher degree of accuracy and a deeper understanding of context, advanced language models available through the **ChatGPT API** can be used. These models are context-aware and better at handling negation, implicit meaning, complex reasoning, and many forms of sarcasm than traditional sentiment analysis methods. However, no model is perfect, and challenging cases involving subtle sarcasm, cultural references, or highly ambiguous language may still lead to misclassification.
