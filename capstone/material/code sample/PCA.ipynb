# Dimensionality Reduction

Dimensionality reduction in machine learning refers to the process of **reducing the number of input variables or features** in a dataset while **retaining as much relevant information as possible**. This is typically done by transforming the high-dimensional data into a lower-dimensional space, where each dimension represents a combination of the original features.

Dimensionality reduction techniques are important for several reason

* **Curse of Dimensionality**: As the number of features increases, the volume of the feature space grows exponentially. This can lead to sparsity of data and increased computational complexity, making algorithms less efficient and more **prone to overfitting**.

* **Computational Efficiency**: High-dimensional data requires more **computational resources** and **time** to process. By reducing the dimensionality of the data, the computational burden can be significantly reduced, allowing for faster training.

* **Improved Model Performance**: Dimensionality reduction can help **improve the performance** of machine learning algorithms by removing redundant or irrelevant features, focusing on the most important aspects of the data, and reducing noise and variability.

* **Visualization**: Lower-dimensional representations of data are **easier to visualize** and **interpret**, especially in two or three dimensions. Dimensionality reduction techniques enable visual exploration of high-dimensional data, helping researchers and practitioners gain insights and identify patterns.


# Principal Component Analysis (PCA) 

Principal Component Analysis (PCA) is a popular dimensionality reduction technique used in machine learning and data analysis. It aims to transform **high-dimensional** data into a **lower-dimensional** space while preserving the **maximum amount of variance** in the original data. Without going deep into the mathematical concepts underlying PCA, let's explore the five steps involved for computing PCA. 

1. **Data Standardization**: PCA typically begins with the standardization of the data to ensure that all features have a **mean of zero** and a **standard deviation of one**. This step is important to give **equal importance to all features** during the analysis.


2. **Covariance Matrix Computation**: PCA calculates the covariance matrix of the standardized data. The covariance matrix provides information about the **relationships between different features** in the dataset. Specifically, it shows how much two features vary together. This is a symmetric matrix, and each element ($i$, $j$) corresponds to the covariance between variables i and j.

    Given a dataset with two features, the covariance matrix for two features $X_1$ and $X_2$ is represented as:

    $\text{cov} =
    \begin{bmatrix}
    \text{cov}(X_1, X_1) & \text{cov}(X_1, X_2) \\
    \text{cov}(X_2, X_1) & \text{cov}(X_2, X_2)
    \end{bmatrix}$

   Or in general, a dataset with $n$ observations (samples), the covariance is calculated as:

     $\text{cov} =
    \frac{1}{n-1}
    \mathbf{X}_c^\top
    \mathbf{X}_c
    \$
   
where:
* $\mathbf{X}_c^\top$ is the centered data matrix (mean subtracted from each feature)
* $n$ is the number of observations

4. **Eigendecomposition**: PCA then performs eigen decomposition on the covariance matrix to obtain the eigenvectors and eigenvalues. **Eigenvectors** represent the **directions of maximum variance** in the data, while **eigenvalues** indicate the **magnitude of variance** along these directions.

    To compute the eigenvalues ($\lambda$), we solve the following equation:
    
    $\text{det}(\text{cov} - \lambda I) = 0$
    
    where, $I$ is the identity matrix, and to compute the eigenvectors, we solve the following equation*:
    
    $\text{cov} \cdot v = \lambda \cdot v$

5. **Selection of Principal Components**: PCA ranks the eigenvectors based on their corresponding eigenvalues in descending order. The eigenvector with the **highest eigenvalue** represents the principal component (PC) that **captures the most variance** in the data. Subsequent eigenvectors represent additional principal components, each capturing progressively less variance*.


6. **Projection onto Principal Components**: Finally, PCA **projects** the original data onto the **principal components** to obtain the **lower-dimensional representation** of the data. This involves multiplying the standardized data matrix by the matrix of selected eigenvectors (PCs), resulting in a new dataset with reduced dimensions*.

**NOTE:** The mathematical computations from steps 2 to 5 are executed in the scikit-learn package using the `PCA()`, `fit()`, and `transform()` functions.


**Example:** The following example first uses KMeans clustering on the Iris dataset. Then it applies PCA on the standardized data to **reduce its dimensionality to 2 from 4** for visualization. Finally, it visualizes the clustered data using a scatter plot.

```python
# Import necessary libraries
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load Iris dataset
iris = load_iris()
X = iris.data

clustering = KMeans(n_clusters = 3).fit(X)
labels = clustering.labels_

# Standardize the data
scaler = StandardScaler()
scaled_X = scaler.fit_transform(X)

# Perform PCA (map 4-dimensional space to 2-dimensional space for plotting purposes)
pca = PCA(n_components=2)  # Specify the number of principal components
pca.fit(scaled_X)  # Fit the PCA model to the scaled data
transformed_data = pca.transform(scaled_X)  # Transform the data to the new feature space

# Visualize the results
plt.figure(figsize=(8, 6))
plt.scatter(transformed_data[:, 0], transformed_data[:, 1], c=labels, cmap='viridis')

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA for Iris data')
plt.show()

eigen_values = pca.explained_variance_ #eigenvalues

eigen_vectors = pca.components_  # Directions of Principal components (i.e., the directions in the original 
# feature space along which the data varies the most (eigenvectors)

variance = pca.explained_variance_ratio_  # Percentage of variance contribution by each of the selected components

print("The first two eigenvalues:", eigen_values)

print("\nDirection of the first two PCs (eignevectors):\n", eigen_vectors)

print("\nVariance ratio for the first two PCs:", variance)

print("\nTotal variance ratio: ", variance.sum())
```

**PCA Results Analysis**

* The direction of the first two principal components (PCs) indicates the linear combinations of the original features that form these components. The first PC influenced by the first, third, and fourth features (higher value indicate higher influence), while the second PC is mainly influenced by the second feature.

* The variance ratio for the first two PCs represents the proportion of variance in the data captured by each PC. The first PC captures approximately 72.96% of the total variance, while the second PC captures around 22.85%. Together, these two PCs capture approximately 95.81% of the total variance in the data, indicating that they provide a comprehensive summary of the dataset's variability.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

# Load the Housing dataset
housing = pd.read_csv('Housing.csv')
print(housing.info())
housing.describe()
```

```python
# Handling negative distance to school
housing["Distance_to_school"] = abs(housing["Distance_to_school"])

sizeThreshold = 3000
medianSize = housing['Size'].median()

# Handling outliers: Fill outliers with the median of the column
housing.loc[housing['Size'] > sizeThreshold, 'Size'] = medianSize

meanSize = housing['Size'].mean()
meanBedroom = housing['Bedrooms'].mean()

# Handling missing data: Fill missing values with the mean of the column
housing['Size'].fillna(meanSize, inplace = True)
housing['Bedrooms'].fillna(meanBedroom, inplace = True)

print(housing.info())
housing.describe()
```

```python
# Standardize the data
scaler = StandardScaler()
scaled_housing = scaler.fit_transform(housing)

# Perform PCA
pca = PCA(n_components=2)  # Specify the number of principal components
pca.fit(scaled_housing)  # Fit the PCA model to the scaled data
transformed_data = pca.transform(scaled_housing)  # Transform the data to the new feature space

# Visualize the results
plt.figure(figsize=(8, 6))
plt.scatter(transformed_data[:, 0], transformed_data[:, 1], color='blue')

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA for Housing Data')
plt.show()

eigen_values = pca.explained_variance_ #eigenvalues

eigen_vectors = pca.components_  # Directions of Principal components (i.e., the directions in the original 
# feature space along which the data varies the most (eigenvectors)

variance = pca.explained_variance_ratio_  # Percentage of variance contribution by each of the selected components

print("The first two eigenvalues:", eigen_values)

print("\nDirection of the first two PCs (eignevectors):", eigen_vectors)

print("\nVariance ratio for the first two PCs:", variance)

print("\nTotal variance ratio: ", variance.sum())
```
