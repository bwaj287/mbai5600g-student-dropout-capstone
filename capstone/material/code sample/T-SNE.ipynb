# t-Distributed Stochastic Neighbor Embedding (t-SNE)

**t-SNE** is a nonlinear dimensionality reduction technique primarily used for **visualizing high-dimensional data in 2D or 3D**. It converts similarities between data points to joint probabilities and tries to minimize the **Kullback-Leibler** divergence between the joint probabilities of the low-dimensional embedding and the high-dimensional data. The key idea is to **preserve the local structure of the data**: points that are **close together** in the original **high-dimensional space** should **remain close together** in the **lower-dimensional representation**.

In general t-SNE:

1. **Compute similarities in the original space**
    
    * For each pair of points, t-SNE converts distances into probabilities.
    * Nearby points get high probabilities; distant points get low probabilities.

2. **Create a low-dimensional map**
    * Points are randomly placed in 2D or 3D.

3. **Match the similarity distributions**
    * t-SNE iteratively moves points so that neighborhood relationships in the low-dimensional space resemble those in the original space.

4. **Use a Student's t-distribution**
    * Instead of a Gaussian distribution in the low-dimensional space, t-SNE uses a heavy-tailed t-distribution.
    * This helps separate clusters and alleviates the "crowding problem."

t-SNE offers several parameters to tune, but among them, the **perplexity** value particularly influences the shape the resulting visualization. Perplexity controls the effective number of neighbors considered for each point. It may be viewed as a **knob** that sets the number of effective nearest neighbors. It is comparable with the number of nearest neighbors $k$ that is employed in many manifold learners. A perplexity of **30** is often a good starting point because it tells t-SNE to preserve neighborhoods consisting of roughly **30 nearby points** around each observation.

The following example use the **digits dataset** with **64 features** to illustrate the influence of perplexity values in t-SNE. In this example we apply apply t-SNE with six different values of perplexity (2, 10, 20, 30, 50, and 100) and visualize the resulting embeddings in separate subplots. By observing the plots, you can see how varying perplexity affects the clustering and global structure of the data in the lower-dimensional space.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
import numpy as np

# Load the digits dataset
digits = load_digits()
X = digits.data
y = digits.target

# Perplexity values
perplexities = [2, 10, 20, 30, 50, 100]

plt.figure(figsize=(15, 10))
for i, perplexity in enumerate(perplexities):
    tsne = TSNE(n_components=2, perplexity=perplexity)
    X_tsne = tsne.fit_transform(X)
    plt.subplot(2, len(perplexities)//2, i+1)
    plt.scatter(X_tsne[:, 0], X_tsne[:, 1], s=1, c=y, cmap='viridis')
    plt.title(f"Perplexity = {perplexity}"  + ", KL ={:.2f}".format(tsne.kl_divergence_))
    for d in range(10):
        # Calculate the center of each digit: only used for ploting purposes
        center = np.mean(X_tsne[y==d], axis=0)
        plt.text(center[0], center[1], d, fontsize = 14, color = 'r')
plt.show()
```
